```python
# Minimal single-cell ATAC-seq visualization using episcanpy

# Install 
    # pip install scanpy matplotlib
import scanpy as sc
import matplotlib.pyplot as plt
```

```python
# Load a small scATAC dataset from episcanpy
data = epi.datasets.synthetic_atac()
data
```

```python
# Basic preprocessing for scATAC data
epi.pp.binarize(data)                     # 0/1 accessibility
epi.pp.tfidf(data)                        # TF-IDF normalisation
epi.tl.lsi(data, n_components=20)         # LSI for dimension reduction

data.obsm["X_lsi"].shape
```

```python
# Use LSI representation to build the graph and compute UMAP
sc.pp.neighbors(data, use_rep="X_lsi")
sc.tl.umap(data)

# Cluster cells
sc.tl.leiden(data, key_added="clusters")

data.obs["clusters"].value_counts()
```

```python
sc.pl.umap(
    data,
    color="clusters",
    frameon=False,
)

# Also save the figure with a simple, hard-coded name
plt.savefig("synthetic_atac_clusters.png", dpi=150, bbox_inches="tight")
```
