```python
# Minimal single-cell ATAC-seq visualization using episcanpy

# Install 
    # pip install scanpy episcanpy matplotlib
import scanpy as sc
import episcanpy.api as epi
import matplotlib.pyplot as plt
```

```python
# Load a very small example scATAC dataset from episcanpy
adata = epi.datasets.synthetic_atac()

adata
```

```python
# Basic preprocessing for scATAC data
epi.pp.binarize(adata)                     # 0/1 accessibility
epi.pp.tfidf(adata)                        # TF-IDF normalisation
epi.tl.lsi(adata, n_components=20)         # LSI for dimension reduction

adata.obsm["X_lsi"].shape
```

```python
# Use LSI representation to build the graph and compute UMAP
sc.pp.neighbors(adata, use_rep="X_lsi")
sc.tl.umap(adata)

# Cluster cells
sc.tl.leiden(adata, key_added="clusters")

adata.obs["clusters"].value_counts()
```
